```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
from IPython.display import display
from scipy import stats
from scipy.optimize import minimize, minimize_scalar
from scipy.special import expit
```

```python
TASK = Path.cwd().parent
ROOT = TASK.parent
INPUT, OUTPUT = TASK / 'input', TASK / 'output'
CONFIG = json.loads((ROOT / 'config.json').read_text())
EPS, Y, SEED = CONFIG['eps'], CONFIG['last_year'], CONFIG['seed']
```

```python
TAG = 'lognormal_grw'
```

```python
pairs = pd.read_csv(INPUT / 'transitions.csv')
DEST = OUTPUT / TAG
DEST.mkdir(exist_ok=True)
```

```python
# normal MLE per transition-start age; variance denominator n
params = pairs.groupby('age', sort=True).dz.agg(mu='mean', sigma=lambda x: x.std(ddof=0), n='size').reset_index()
params.to_csv(DEST / 'parameters.csv', index=False)
display(params)
```
