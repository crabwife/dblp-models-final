```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import display
```

```python
TASK = Path.cwd().parent.parent
ROOT = TASK.parent
INPUT, OUTPUT = TASK / 'input', TASK / 'output'
CONFIG = json.loads((ROOT / 'config.json').read_text())
TAG = 'lognormal_grw_s'
```

```python
pairs = pd.read_csv(INPUT / 'transitions.csv')
DEST = OUTPUT / TAG
DEST.mkdir(exist_ok=True)
```

```python
rows = []
# stage-specific normal MLE on observed log-offset increments
for label, start, end in CONFIG['fit_stages']:
    increments = pairs.loc[pairs.age.between(start,end),'dz']
    rows.append({'stage':label, 'start':start, 'end':end, 'mu':increments.mean(),
                 'sigma':increments.std(ddof=0), 'n':len(increments)})
params = pd.DataFrame(rows)
```

```python
params.to_csv(DEST / 'parameters.csv', index=False)
display(params)
```
