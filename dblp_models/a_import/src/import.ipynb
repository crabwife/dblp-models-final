```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
from IPython.display import display
import hashlib
```

```python
TASK = Path.cwd().parent
ROOT = TASK.parent
INPUT, OUTPUT = TASK / 'input', TASK / 'output'
CONFIG = json.loads((ROOT / 'config.json').read_text())
EPS, Y, SEED = CONFIG['eps'], CONFIG['last_year'], CONFIG['seed']
```

```python
df = pd.read_csv(INPUT / 'adjusted_productivity.csv')
```

```python
# source contract; checked once at entry
required = ['dblp', 'phd_year', 'year', 'CareerAge', 'pubs_adj']
assert set(required).issubset(df.columns)
assert df[['dblp', 'year', 'CareerAge', 'pubs_adj']].notna().all().all()
assert np.isfinite(df[['year', 'CareerAge', 'pubs_adj']]).all().all()
assert df.pubs_adj.ge(0).all()
assert not df.duplicated(['dblp', 'phd_year', 'CareerAge']).any()
assert df.year.eq(df.year.astype(int)).all() and df.CareerAge.eq(df.CareerAge.astype(int)).all()
```

```python
df.to_csv(OUTPUT / 'source.csv', index=False)
source = pd.DataFrame([{'filename':'adjusted_productivity.csv', 'rows':len(df),
                       'sha256':hashlib.sha256((INPUT / 'adjusted_productivity.csv').read_bytes()).hexdigest()}])
source.to_csv(OUTPUT / 'source_manifest.csv', index=False)
display(source)
```
