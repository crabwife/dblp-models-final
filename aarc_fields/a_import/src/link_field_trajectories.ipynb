```python
import os
from pathlib import Path
```

```python
TASK = Path.cwd().parent
DEFAULT_SOURCE = Path.home()/'Desktop/NSF 26-256/b_trajectorize/output/fields'
SOURCE = Path(os.environ.get('AARC_FIELDS_DIR', DEFAULT_SOURCE)).expanduser().resolve()
INPUT = TASK/'input/fields'
OUTPUT = TASK/'output/fields'
```

```python
# keep restricted trajectories in the source project
if not SOURCE.is_dir():
    raise FileNotFoundError(f'AARC field directory not found: {SOURCE}. Set AARC_FIELDS_DIR to its path.')

if INPUT.is_symlink() and INPUT.resolve() != SOURCE:
    INPUT.unlink()
if not INPUT.is_symlink():
    INPUT.symlink_to(SOURCE, target_is_directory=True)
if not OUTPUT.is_symlink():
    OUTPUT.symlink_to('../input/fields', target_is_directory=True)
```
