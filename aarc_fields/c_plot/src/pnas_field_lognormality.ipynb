```python
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator, StrMethodFormatter
import pandas as pd
import seaborn as sns
```

```python
TASK = Path.cwd().parent
INPUT, OUTPUT = TASK/'input', TASK/'output/pnas'
OUTPUT.mkdir(parents=True, exist_ok=True)
sns.set_theme(style='ticks', context='paper', font_scale=1,
              rc={'font.size':8, 'axes.labelsize':8, 'xtick.labelsize':7,
                  'ytick.labelsize':7, 'pdf.fonttype':42, 'ps.fonttype':42,
                  'savefig.dpi':400})
```

```python
summary = pd.read_csv(INPUT/'aarc_field_lognormality.csv').dropna(subset=['qq_r'])
if summary.empty:
    raise ValueError('No field has a Q–Q correlation; see aarc_field_lognormality.csv for field statuses')
```

```python
fig, ax = plt.subplots(figsize=(3.35, 2.75), layout='constrained')
positions = range(len(summary))
sns.scatterplot(x=summary.qq_r, y=list(positions), ax=ax,
                color=sns.color_palette('colorblind')[2], s=23, linewidth=0, zorder=3)
ax.set_yticks(list(positions), labels=summary.label)
ax.yaxis.tick_right()
ax.tick_params(axis='y', length=0, pad=5)
ax.grid(axis='x', color='.9', linewidth=.6)
ax.set(xlabel=r'Lognormal Q–Q correlation, $r$', ylabel='', ylim=(-.7, len(summary)-.3),
       xlim=(max(0, summary.qq_r.min()-.006), 1.002))
ax.xaxis.set_major_locator(MaxNLocator(nbins=4, steps=[1, 2, 2.5, 5, 10]))
ax.xaxis.set_major_formatter(StrMethodFormatter('{x:.3f}'))
sns.despine(ax=ax, left=True)
for extension in ['pdf', 'png']:
    fig.savefig(OUTPUT/f'pnas_aarc_field_lognormality.{extension}', facecolor='white')
plt.show()
plt.close(fig)
```
